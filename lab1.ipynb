```python
import heapq

def reconstruct_path(prev, start, end):
    if prev[end] == -1 and end != start:
        return []  # пути нет
    path = []
    cur = end
    while cur != -1:
        path.append(cur)
        cur = prev[cur]
    path.reverse()
    if not path or path[0] != start:
        return []
    return path


# --- Вариант 1: матрица смежности ---
def dijkstra_matrix(adj_matrix, start):
    n = len(adj_matrix)
    dist = [float('inf')] * n
    prev = [-1] * n
    dist[start] = 0

    pq = [(0, start)]
    visited = [False] * n

    while pq:
        d, u = heapq.heappop(pq)
        if visited[u]:
            continue
        visited[u] = True

        for v in range(n):
            weight = adj_matrix[u][v]
            # Пропускаем отсутствующие рёбра: None или inf
            if weight is None or weight == float('inf'):
                continue
            if dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))

    return dist, prev


# --- Вариант 2: список смежности ---
def dijkstra_adj_list(adj, start):
    n = len(adj)
    dist = [float('inf')] * n
    prev = [-1] * n
    dist[start] = 0

    pq = [(0, start)]
    visited = [False] * n

    while pq:
        d, u = heapq.heappop(pq)
        if visited[u]:
            continue
        visited[u] = True

        for v, weight in adj[u]:
            if dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))

    return dist, prev


# ==========================================
# Примеры для Colab (можно менять прямо тут)
# ==========================================

# Пример 1: матрица смежности
adj_matrix = [
    [None, 10,  3, None, None],
    [None, None, 1,  2,  None],
    [None,  4, None,  8,   2],
    [None, None, None, None,  7],
    [None, None, None,  9, None]
]
start_node = 0

dist_mat, prev_mat = dijkstra_matrix(adj_matrix, start_node)
print("=== Матрица смежности ===")
print(f"Расстояния от вершины {start_node}: {dist_mat}")
for t in range(len(dist_mat)):
    path = reconstruct_path(prev_mat, start_node, t)
    print(f"Путь до {t}: {path} (длина {dist_mat[t] if dist_mat[t] != float('inf') else '∞'})")

# Пример 2: список смежности
adj_list = [
    [(1, 10), (2, 3)],          # 0
    [(2, 1), (3, 2)],          # 1
    [(1, 4), (3, 8), (4, 2)],  # 2
    [(4, 7)],                  # 3
    [(3, 9)]                   # 4
]

dist_adj, prev_adj = dijkstra_adj_list(adj_list, start_node)
print("\n=== Список смежности ===")
print(f"Расстояния от вершины {start_node}: {dist_adj}")
for t in range(len(dist_adj)):
    path = reconstruct_path(prev_adj, start_node, t)
    print(f"Путь до {t}: {path} (длина {dist_adj[t] if dist_adj[t] != float('inf') else '∞'})")
```
